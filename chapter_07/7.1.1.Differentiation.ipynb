# 기울기 

- 기울기는 x가 변함에 따라 y가 얼마나 변하는지 말해준다. 
- 기울기 m은 x와 y의 비례 상수 ( proportionality constant )
- 절편 b는 상수 오프셋  

## 할선 ( secant line ) 

- 곡선의 두 점을 지나는 직선

## 접선 ( tangent line )

- 한점에서 곡선과 접하는 직선 


## 정류점 ( stationary point )

- 접선의 기울기가 0인 점들 

## 도함수의 극한 

도함수의 극한은 “할선의 기울기”를 점점 가까이 보내서 “접선의 기울기”를 얻는 과정입니다.

정의:
$$
f'(x)=\lim_{h\to 0}\frac{f(x+h)-f(x)}{h}
$$

특정 점 \(a\)에서의 도함수:
$$
f'(a)=\lim_{h\to 0}\frac{f(a+h)-f(a)}{h}
$$

좋은 질문입니다. 핵심은 이거예요.

- \(\Delta\): 유한한 변화량 (finite change)
- \(d\): 무한소 변화량 (infinitesimal change, 미분에서의 이상적 한계 개념)

관계는 “\(\Delta\)를 점점 0으로 보내면 \(d\) 개념으로 간다”로 이해하면 됩니다.

예시로 \(y=f(x)\)일 때:

1. 평균변화율(할선 기울기)

$$
\frac{\Delta y}{\Delta x} =
\frac{f(x+\Delta x)-f(x)}{\Delta x}
$$

2. 순간변화율(접선 기울기, 도함수)

$$
\frac{dy}{dx}
=
\lim_{\Delta x\to 0}\frac{\Delta y}{\Delta x}
$$

3. 작은 변화에서의 근사

$$
\Delta y \approx f'(x)\,\Delta x
$$

$$
dy = f'(x)\,dx
$$

정리:
- \(\Delta\)는 실제로 “얼마나 바뀌었는지” 측정할 때 사용
- \(d\)는 극한을 통과한 “순간적인 변화 법칙”을 나타낼 때 사용
- 그래서 미적분에서는 \(\Delta\to d\)가 아니라, \(\Delta\)의 극한 과정으로 \(d\)가 정의됩니다.

훌륭한 질문입니다, 미래의 선구자님! 인공지능의 심연으로 한 걸음 더 들어가기 위해 **몫의 미분법, 로그와 지수함수의 확장, 그리고 삼각함수**까지 호기심을 넓히신 것에 스승으로서 큰 기쁨을 느낍니다.

단순히 x의 거듭제곱을 넘어서, 이 복잡한 초월함수(Transcendental Functions)들이 왜 탄생했고, 최첨단 AI 모델(특히 대형 언어 모델인 LLM)에서 어떻게 숨 쉬고 있는지 그 비밀을 모두 밝혀드리겠습니다.

### 1. 몫의 미분법 (Quotient Rule): $\frac{f(x)}{g(x)}$

나눗셈 형태로 된 함수를 미분하는 방법입니다.

* **왜 탄생했는가?** 비율, 평균, 혹은 확률처럼 무언가를 '나누어' 정규화(Normalization)하는 과정의 변화율을 측정하기 위해 고안되었습니다.
* **AI에서의 쓰임새:** AI가 여러 선택지 중 하나의 정답을 고를 때(예: 사진을 보고 고양이인지 강아지인지 맞출 때), 출력값을 0에서 1 사이의 '확률'로 변환해 주는 **소프트맥스(Softmax)** 함수가 바로 이 분수 형태를 띱니다. 소프트맥스 층의 역전파(BP)를 계산할 때 몫의 미분법이 핵심적으로 사용됩니다.

**정의 및 공식:**


$$y = \frac{f(x)}{g(x)} \quad (단, g(x) \neq 0)$$

$$y' = \frac{f'(x)g(x) - f(x)g'(x)}{\{g(x)\}^2}$$

**증명 (곱의 미분법 활용):**
이 증명은 극한보다 훨씬 우아한 방식을 사용합니다. $y = \frac{f(x)}{g(x)}$의 양변에 $g(x)$를 곱하면,


$$y \cdot g(x) = f(x)$$


양변을 $x$에 대해 미분합니다 (곱의 미분법 $(uv)' = u'v + uv'$ 적용).


$$y' \cdot g(x) + y \cdot g'(x) = f'(x)$$


우리가 구하고 싶은 것은 $y'$이므로 식을 정리합니다.


$$y' \cdot g(x) = f'(x) - y \cdot g'(x)$$


여기에 다시 $y = \frac{f(x)}{g(x)}$를 대입합니다.


$$y' \cdot g(x) = f'(x) - \frac{f(x)}{g(x)} \cdot g'(x)$$

$$y' \cdot g(x) = \frac{f'(x)g(x) - f(x)g'(x)}{g(x)}$$


양변을 다시 $g(x)$로 나누면 공식이 증명됩니다.


$$y' = \frac{f'(x)g(x) - f(x)g'(x)}{\{g(x)\}^2}$$

### 2. 다양한 지수함수와 로그함수의 미분법

* **왜 탄생했는가?** 지수함수는 세포 분열이나 복리 이자처럼 '폭발적으로 성장하는 양'을 묘사하기 위해, 로그함수는 천문학자들이 곱셈을 덧셈으로 바꾸어 거대한 숫자의 계산을 쉽게 하기 위해(존 네이피어) 발명되었습니다.
* **AI에서의 쓰임새:** * **로그 (Log):** AI가 얼마나 정답을 확신하는지 평가하는 **교차 엔트로피 손실(Cross-Entropy Loss)** 함수에 쓰입니다. 매우 작은 확률값들을 계속 곱하면 0으로 사라져버리는 문제(언더플로우)를 방지하고, 곱셈 연산을 덧셈으로 바꾸어 컴퓨터의 연산 속도를 극대화합니다.
* **지수 (Exp):** 앞서 말한 소프트맥스 함수나 시그모이드(Sigmoid) 활성화 함수에서, 어떤 실수든 0보다 큰 양수로 뻥튀기하여 확률로 만들기 위해 사용됩니다.



**정의 및 공식:**

* 일반 지수함수: $f(x) = a^x \implies f'(x) = a^x \ln a$
* 자연 로그함수: $f(x) = \ln x \implies f'(x) = \frac{1}{x}$
* 밑이 a인 로그: $f(x) = \log_a x \implies f'(x) = \frac{1}{x \ln a}$

**로그함수 $f(x) = \ln x$ 의 미분 증명:**
로그의 정의에 의해, $y = \ln x$ 이면 $x = e^y$ 입니다. 양변을 $x$에 대하여 미분하면 (음함수 미분법),


$$1 = e^y \cdot y'$$

$$y' = \frac{1}{e^y}$$


$e^y = x$ 이므로, 결국 $y' = \frac{1}{x}$ 가 됩니다. 아주 아름답고 간결한 결과지요!

### 3. 삼각함수의 미분법 (Sin, Cos, Tan)

* **왜 탄생했는가?** 고대 그리스 시절, 별의 위치를 측정하고 건축물을 짓기 위해 각도와 변의 길이 비율을 연구하며 탄생했습니다. 이후 파동(소리, 빛)이나 진동 같은 주기적 현상을 수학적으로 모델링하는 데 쓰였습니다.
* **AI에서의 쓰임새:** 챗GPT(ChatGPT)를 만든 핵심 기술인 **트랜스포머(Transformer)** 구조에서 위치 인코딩(Positional Encoding)에 사용됩니다. 언어 모델은 단어를 동시에 통째로 입력받기 때문에 단어의 '순서(Time-step)'를 모릅니다. 이때 삼각함수의 주기성(올라갔다 내려왔다 하는 성질)을 이용해 각 단어에 위치 정보를 고유한 파동 형태로 부여합니다.

**정의 및 공식:**

* $f(x) = \sin x \implies f'(x) = \cos x$
* $f(x) = \cos x \implies f'(x) = -\sin x$
* $f(x) = \tan x \implies f'(x) = \sec^2 x$ (탄젠트는 $\frac{\sin x}{\cos x}$이므로 앞서 배운 **몫의 미분법**으로 쉽게 증명됩니다!)

**$\sin x$ 의 미분 증명:**
도함수의 정의와 삼각함수의 덧셈정리 $\sin(x+h) = \sin x \cos h + \cos x \sin h$ 를 사용합니다.


$$f'(x) = \lim_{h \to 0} \frac{\sin(x+h) - \sin x}{h}$$

$$= \lim_{h \to 0} \frac{\sin x \cos h + \cos x \sin h - \sin x}{h}$$


$\sin x$ 와 $\cos x$ 로 항을 묶어줍니다.


$$= \lim_{h \to 0} \left( \sin x \frac{\cos h - 1}{h} + \cos x \frac{\sin h}{h} \right)$$


수학적으로 $\lim_{h \to 0} \frac{\cos h - 1}{h} = 0$ 이고, $\lim_{h \to 0} \frac{\sin h}{h} = 1$ 로 수렴함이 알려져 있습니다. 따라서 첫 번째 항은 사라지고 두 번째 항만 남아 $\cos x$ 가 됩니다.

### 5. Python 코드 예제: 자동 미분으로 확인하는 초월함수

이론으로 배운 내용들을 PyTorch의 자동 미분 엔진이 어떻게 계산하는지 확인해 볼 수 있는 Jupyter Notebook 코드입니다.


이 코드를 실행하시면, 우리가 손으로 힘들게 증명했던 $e^x$ 의 도함수가 자기 자신과 완전히 겹치는 모습, $\sin x$ 의 도함수가 $\cos x$ 형태로 반 발짝 먼저 파동 치는 모습 등을 AI 프레임워크가 알아서 계산해 그려주는 것을 볼 수 있습니다.

```python

# [셀 1: 라이브러리 로드]
import torch
import matplotlib.pyplot as plt

# [셀 2: 다양한 함수의 미분값 확인하기]
def explore_derivatives():
    # 1. 미분을 추적할 텐서(x) 생성. -2 부터 2까지 100개의 점을 생성합니다.
    x = torch.linspace(-2, 2, 100, requires_grad=True)
    
    # 2. 분석할 함수들을 딕셔너리로 정의 (몫의 미분, 지수, 로그, 삼각함수)
    # 참고: log는 양수에서만 정의되므로 x_log를 따로 만듭니다.
    x_log = torch.linspace(0.1, 4, 100, requires_grad=True)
    
    functions = {
        'Quotient (Softmax-like): f(x)=x/(x^2+1)': x / (x**2 + 1),
        'Exponential: f(x)=e^x': torch.exp(x),
        'Logarithmic: f(x)=ln(x)': torch.log(x_log),
        'Trigonometric: f(x)=sin(x)': torch.sin(x)
    }

    plt.figure(figsize=(12, 10))
    
    # 3. 각 함수에 대해 미분을 수행하고 그래프를 그립니다.
    for i, (name, y) in enumerate(functions.items()):
        # y값들의 합에 대해 backward()를 호출하여 각 x 지점에서의 기울기를 계산합니다.
        # (스칼라 값으로 만들기 위해 sum()을 사용합니다)
        y.sum().backward(retain_graph=True)
        
        # 사용하는 x 텐서 결정 (log만 x_log 사용)
        current_x = x_log if 'ln' in name else x
        
        # 시각화 (detach()를 사용하여 기울기 추적에서 분리한 값을 넘파이 배열로 변환)
        plt.subplot(2, 2, i+1)
        plt.plot(current_x.detach().numpy(), y.detach().numpy(), label='f(x) Original', color='blue', linewidth=2)
        plt.plot(current_x.detach().numpy(), current_x.grad.detach().numpy(), label="f'(x) Derivative", color='red', linestyle='--')
        
        plt.title(name)
        plt.xlabel('x')
        plt.ylabel('y')
        plt.legend()
        plt.grid(True)
        
        # 다음 계산을 위해 기울기(grad)를 0으로 초기화합니다.
        current_x.grad.zero_()

    plt.tight_layout()
    plt.show()

# 실행
explore_derivatives()
```
