# Projection(사영)

### 정사영(Vector Projection)의 개념과 유도

벡터 $\mathbf{a}$를 다른 벡터 $\mathbf{b}$ 위로 정사영시킨 벡터를 $\text{proj}_{\mathbf{b}}\mathbf{a}$ 라고 표기합니다. 
이 벡터는 두 가지 요소로 이루어집니다.
1. **방향**: 벡터 $\mathbf{b}$와 같은 방향을 가집니다. 즉, $\mathbf{b}$의 단위 벡터 $\hat{\mathbf{b}}$ 의 방향입니다.
2. **크기(스칼라 투영)**: 직각삼각형의 삼각비에 의해 $\|\mathbf{a}\| \cos\theta$ 가 됩니다.

따라서 정사영 벡터 본래의 식은 수식으로 다음과 같습니다.
$$ \text{proj}_{\mathbf{b}}\mathbf{a} = (\|\mathbf{a}\| \cos\theta) \hat{\mathbf{b}} $$

#### 유도(Proof) 과정
미리 배운 내적의 기하학적 정의($\mathbf{a} \cdot \mathbf{b} = \|\mathbf{a}\| \|\mathbf{b}\| \cos\theta$)와 단위 벡터 공식($\hat{\mathbf{b}} = \frac{\mathbf{b}}{\|\mathbf{b}\|}$)을 결합하여 식을 풀어냅니다.

1. 먼저 스칼라 투영 길이인 $\|\mathbf{a}\| \cos\theta$ 를 내적 식으로 바꿉니다.
$$ \|\mathbf{a}\| \cos\theta = \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{b}\|} $$

2. 단위 벡터 $\hat{\mathbf{b}}$ 를 대입합니다.
$$ \hat{\mathbf{b}} = \frac{\mathbf{b}}{\|\mathbf{b}\|} $$

3. 위 두 식을 곱하여 최종 정사영 벡터를 구합니다.
$$ \text{proj}_{\mathbf{b}}\mathbf{a} = \left( \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{b}\|} \right) \left( \frac{\mathbf{b}}{\|\mathbf{b}\|} \right) = \frac{\mathbf{a} \cdot \mathbf{b}}{\|\mathbf{b}\|^2} \mathbf{b} $$

4. (추가) 벡터 자신과의 내적은 크기의 제곱 ($\mathbf{b} \cdot \mathbf{b} = \|\mathbf{b}\|^2$) 이므로 다음과 같이 완벽한 내적 형태로도 표현할 수 있습니다.
$$ \text{proj}_{\mathbf{b}}\mathbf{a} = \left( \frac{\mathbf{a} \cdot \mathbf{b}}{\mathbf{b} \cdot \mathbf{b}} \right) \mathbf{b} $$

```python
import numpy as np
import matplotlib.pyplot as plt

# 두 벡터 정의
a = np.array([2, 5])
b = np.array([6, 2])

# Projection 공식 적용: (a dot b / b dot b) * b
# np.dot 을 이용한 내적 계산
proj_b_a = (np.dot(a, b) / np.dot(b, b)) * b

print(f"Vector a: {a}")
print(f"Vector b: {b}")
print(f"Projection of a onto b: {proj_b_a}")

# --- 결과 시각화 ---
plt.figure(figsize=(7, 7))

# 원점, 벡터 a, 벡터 b 그리기
plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue', label='Vector a')
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='red', label='Vector b')

# 투영된 벡터 proj_b_a 그리기 (초록색 두꺼운 선)
plt.quiver(0, 0, proj_b_a[0], proj_b_a[1], angles='xy', scale_units='xy', scale=1, color='green', linewidth=3, label='Proj of a onto b')

# a에서 떨어지는 수선의 발(점선) 그리기
plt.plot([a[0], proj_b_a[0]], [a[1], proj_b_a[1]], 'k--', label='Orthogonal Line')

plt.xlim(-1, 7)
plt.ylim(-1, 7)
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)
plt.grid(True)
plt.legend()
plt.title('Vector Projection Visualization')
plt.show()
```
