# 코사인 유사도


선택하신 내적의 3가지 주요 성질에 대한 수학 공식은 다음과 같습니다.

**1. 벡터 자신과의 내적은 크기의 제곱이다.**
$$ \mathbf{a} \cdot \mathbf{a} = \|\mathbf{a}\|^2 $$

**2. 내적 연산은 가환적(commutative)이다. (교환 법칙 성립)**
$$ \mathbf{a} \cdot \mathbf{b} = \mathbf{b} \cdot \mathbf{a} $$

**3. 분배 법칙도 성립한다.**
$$ \mathbf{a} \cdot (\mathbf{b} + \mathbf{c}) = \mathbf{a} \cdot \mathbf{b} + \mathbf{a} \cdot \mathbf{c} $$

```python
import numpy as np

a = np.array([1,2,3,4])
b = np.array([5,6,7,8])

def inner(a, b):
    s = 0.0
    for i in range(len(a)):
        s += a[i] * b[i]
        return s

inner(a,b)

print(r"(a*b).sum()=", (a*b).sum())
print(r"np.dot(a,b)=", np.dot(a,b))
A = np.sqrt(np.dot(a, a))
B = np.sqrt(np.dot(b, b))
t = np.arccos(np.dot(a,b)/(A*B))
print(r"t*(180/np.pi)=", t*(180/np.pi))
print(r"np.degrees(t)=", np.degrees(t))
print(r"np.rad2deg(t)=", np.rad2deg(t))
```

```python
# 3 diension array 
a = np.array([1,0,0])
b = np.array([0,1,0])
A = np.sqrt(np.dot(a, a))
B = np.sqrt(np.dot(b, b))
t = np.arccos(np.dot(a,b)/(A*B))
print(r"t*(180/np.pi)=", t*(180/np.pi))
print(r"np.degrees(t)=", np.degrees(t))
print(f"np.rad2deg(t)={np.rad2deg(t)}")
print(r"np.dot(a,b)=", np.dot(a,b))
```
