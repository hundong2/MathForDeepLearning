# 가설검정

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# 1. 가상의 AI 모델 성능 데이터 생성 (정규분포를 따른다고 가정)
np.random.seed(42)
# 모델 A: 평균 80, 표준편차 5, 샘플 30개
model_A = np.random.normal(loc=80, scale=5, size=30)
# 모델 B: 평균 84, 표준편차 5, 샘플 30개
model_B = np.random.normal(loc=84, scale=5, size=30)

# 2. T-검정 수행
t_stat, p_val = stats.ttest_ind(model_A, model_B)

# 3. 시각화 (Visualizing the Distributions)
plt.figure(figsize=(10, 6))

# 데이터의 밀도 함수(KDE) 그리기 대용으로 정규분포 곡선 플롯
x = np.linspace(60, 100, 1000)
pdf_A = stats.norm.pdf(x, np.mean(model_A), np.std(model_A, ddof=1))
pdf_B = stats.norm.pdf(x, np.mean(model_B), np.std(model_B, ddof=1))

plt.plot(x, pdf_A, label=f'Model A (Mean: {np.mean(model_A):.1f})', color='blue')
plt.plot(x, pdf_B, label=f'Model B (Mean: {np.mean(model_B):.1f})', color='red')
plt.fill_between(x, pdf_A, alpha=0.3, color='blue')
plt.fill_between(x, pdf_B, alpha=0.3, color='red')

plt.title(f'AI Model Performance Comparison\nT-statistic: {t_stat:.2f} | P-value: {p_val:.4f}')
plt.xlabel('Accuracy Score')
plt.ylabel('Density (Probability)')
plt.axvline(np.mean(model_A), color='blue', linestyle='dashed', linewidth=1)
plt.axvline(np.mean(model_B), color='red', linestyle='dashed', linewidth=1)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

# Z-검정 (Z-test)

**Z-검정**은 표본의 평균이 모집단의 평균과 일치하는지, 혹은 두 독립된 표본 집단의 평균이나 비율에 통계적으로 유의미한 차이가 있는지를 검증하는 가설 검정 방법입니다.
주로 표본의 크기($n$)가 충분히 크거나(보통 $n \ge 30$, 중심극한정리를 따를 때), **모집단의 분산($\sigma^2$)을 이미 알고 있을 때** 사용합니다. 

현업(예: IT, 마케팅)에서는 **A/B 테스트의 클릭률(CTR)이나 전환율(CVR) 차이를 검증할 때 두 비율의 차이에 대한 Z-검정(Proportions Z-test)**이 매우 빈번하게 사용됩니다.

### Z-검정의 주요 종류와 수식

1. **단일 표본 Z-검정 (One-sample Z-test)**
   - 하나의 표본 평균이 알려진 모집단 평균과 같은지 비교합니다.
   - $Z = \frac{\bar{x} - \mu}{\sigma / \sqrt{n}}$

2. **독립 표본 Z-검정 (Two-sample Z-test)**
   - 두 독립된 집단의 표본 평균에 차이가 있는지 비교합니다.
   - $Z = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\frac{\sigma_1^2}{n_1} + \frac{\sigma_2^2}{n_2}}}$

3. **비율 Z-검정 (Two-proportion Z-test)**
   - 두 집단이 성공할 확률(비율, 예: 클릭률)에 차이가 있는지 검정합니다. 아래의 파이썬 코드 예시에서 사용한 방법입니다.
   - $Z = \frac{\hat{p}_1 - \hat{p}_2}{\sqrt{\hat{p}(1-\hat{p})(\frac{1}{n_1} + \frac{1}{n_2})}}$  $(단, \hat{p}는 합동비율)$

### T-검정 vs Z-검정 요약
- **T-검정 (T-test)**: 모집단의 분산을 모를 때, 특히 표본의 크기가 작을 때 ($n < 30$) 주로 사용합니다. (t분포 사용)
- **Z-검정 (Z-test)**: 표본 크기가 충분히 크거나 ($n \ge 30$) 모집단 분산을 알고 있을 때 사용합니다. (표준정규분포 사용)

### 가설 설정 및 해석
* **귀무가설($H_0$)**: 통계적으로 두 집단의 차이가 **없다** (예컨대 $A$버튼과 $B$버튼의 클릭률은 같다).
* **대립가설($H_1$)**: 통계적으로 두 집단의 차이가 **있다** (유의미한 개선 혹은 저하가 있다).
* **해석 가이드**: 코드 결과의 `p-value`가 정해진 유의수준(일반적으로 $\alpha = 0.05$)보다 작으면 귀무가설을 기각합니다. 즉, "두 집단의 차이가 우연히 발생한 것이 아니며 유의미하다"고 판단합니다.

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# A집단(기존): 1000명 중 120명 클릭 (12%)
# B집단(신규): 1000명 중 150명 클릭 (15%)
successes = np.array([120, 150]) # x_A, x_B
trials = np.array([1000, 1000])  # n_A, n_B

# Z-검정 수행
z_stat, p_value = proportions_ztest(count=successes, nobs=trials)

print(f"Z-통계량: {z_stat:.4f}")
print(f"p-값: {p_value:.4f}")
```
