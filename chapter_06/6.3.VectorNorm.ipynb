**벡터 노름(Vector Norm)**은 벡터의 '크기' 또는 '길이'를 수치화하여 측정하는 함수입니다. 임의의 벡터가 원점 $0$에서 얼마나 멀리 떨어져 있는지 그 복잡한 다차원상의 거리를 하나의 실수값으로 표현합니다.

수학적으로 노름 $||\mathbf{x}||$은 다음 세 가지 조건을 만족해야 합니다:
1. **비음수성(Non-negativity)**: $||\mathbf{x}|| \ge 0$ 이며, $||\mathbf{x}|| = 0$인 것은 $\mathbf{x} = \mathbf{0}$일 때뿐입니다.
2. **동차성(Homogeneity)**: 임의의 실수 $a$에 대해 $||a\mathbf{x}|| = |a| \cdot ||\mathbf{x}||$입니다.
3. **삼각부등식(Triangle Inequality)**: $||\mathbf{x} + \mathbf{y}|| \le ||\mathbf{x}|| + ||\mathbf{y}||$입니다.

### 대표적인 패턴
딥러닝과 기계 학습에서 흔히 사용되는 노름의 종류는 다음과 같습니다:

* **$L_1$ 노름 (맨해튼 노름)**: 벡터 성분들의 절대값 합입니다. 가중치의 희소성(Sparsity)을 유도할 때(L1 규제) 주로 사용됩니다.
  $$||\mathbf{x}||_1 = \sum_{i=1}^{n} |x_i|$$
* **$L_2$ 노름 (유클리드 노름)**: 기하학적인 직선 거리입니다. 가중치 감쇠(L2 규제, Weight Decay)나 일반적인 거리 계산에 가장 널리 쓰입니다.
  $$||\mathbf{x}||_2 = \sqrt{\sum_{i=1}^{n} x_i^2}$$
* **$L_\infty$ 노름 (최대 노름)**: 성분의 절대값 중 최대값입니다.
  $$||\mathbf{x}||_\infty = \max_{i} (|x_i|)$$

현재 열려 있는 6.3.VectorNorm.ipynb에서 벡터 노름에 대한 실습 코드와 실질적인 예시를 상세히 확인해보실 수 있습니다.

```python
import torch

# 1. 임의의 3차원 벡터 생성
# 예를 들어, [가로로 3, 세로로 -4, 높이로 0] 이동한 벡터
v = torch.tensor([3.0, -4.0, 0.0], dtype=torch.float32)

# 2. L1 노름 계산 (절댓값의 합: |3| + |-4| + |0| = 7)
l1_norm = torch.linalg.vector_norm(v, ord=1)

# 3. L2 노름 계산 (제곱합의 루트: sqrt(3^2 + (-4)^2) = 5)
l2_norm = torch.linalg.vector_norm(v, ord=2)

# 4. L-infinity 노름 계산 (최댓값: max(3, 4, 0) = 4)
linf_norm = torch.linalg.vector_norm(v, ord=float('inf'))

print(f"원본 벡터 v: {v.tolist()}")
print(f"L1 노름 (맨해튼 거리)  : {l1_norm.item():.2f}")
print(f"L2 노름 (유클리드 거리): {l2_norm.item():.2f}")
print(f"L∞ 노름 (최대 거리)   : {linf_norm.item():.2f}")
```
