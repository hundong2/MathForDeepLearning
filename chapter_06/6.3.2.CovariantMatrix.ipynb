# 공분산 행렬

```python
from sklearn import datasets 

iris = datasets.load_iris()
X = iris.data[:5]
X
```

```python
import numpy as np
print(np.cov(X, rowvar=False)) #np.cov는 주어진 행렬의 각행이 변수, 열은 그 변수의 관측 값이라고 가정한다. rowvar=False로 설정하면 각 열이 변수, 행은 관측 값이 된다.
```

```python
print(np.std(X, axis=0)**2) #각 열의 표준편차의 제곱은 분산과 같다. np.cov는 편향되지 않은 분산을 계산하기 때문에 n-1로 나누지만, np.std는 편향된 분산을 계산하기 때문에 n으로 나눈다. 따라서 np.std(X, axis=0)**2는 np.cov(X, rowvar=False)의 대각 원소와 다를 수 있다.
```

## 공분산 행렬

- 0번 열과 1번 열이 얼마나 다른지 아는 것이 데이터를 파악하는데 도움이 될 수 있다. 
- 그런 분산을 파악하는 순단이 `공분산 행렬(covariance matrix)`

**공분산 행렬(Covariance Matrix)**의 각 성분(원소)을 계산하는 식은 다음과 같습니다.

### 1. 모집단 공분산 행렬의 성분 식
서로 다른 확률 변수 $X_i$와 $X_j$ 사이의 공분산 성분 $\Sigma_{ij}$는 다음과 같이 정의됩니다.

$$\Sigma_{ij} = \text{Cov}(X_i, X_j) = E[(X_i - E[X_i])(X_j - E[X_j])]$$

여기서 $E[\cdot]$는 기댓값(평균)을 의미합니다.

---

### 2. 표본 공분산 행렬의 성분 식
실제 데이터 분석이나 머신러닝에서 자주 사용하는 표본 데이터셋에 대한 표본 공분산 행렬의 성분 $q_{ij}$는 다음과 같이 계산합니다. (데이터 개수 $N$)

$$q_{ij} = \frac{1}{N - 1} \sum_{k=1}^{N} (x_{ki} - \bar{x}_i)(x_{kj} - \bar{x}_j)$$

여기서 각 기호의 정의는 다음과 같습니다.
- $N$: 전체 표본(데이터 포인트)의 개수
- $x_{ki}$: $k$번째 데이터의 $i$번째 특성(열) 값
- $x_{kj}$: $k$번째 데이터의 $j$번째 특성(열) 값
- $\bar{x}_i$: $i$번째 특성(열)의 표본 평균
- $\bar{x}_j$: $j$번째 특성(열)의 표본 평균
